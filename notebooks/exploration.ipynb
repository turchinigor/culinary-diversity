```python
import pandas as pd
import openpyxl
import geopandas as gpd
import requests
import shapely
import folium
import plotly as plt
import osmnx
import os
from pathlib import Path
from os.path import join
```

```python
DATA_PATH = Path(os.getcwd()).parent / "data"
print(DATA_PATH)
```

```python
city = "Chisinau, Moldova"
places = ["restaurant", "fast_food", "pub", "bar", "cafe"]
```

```python
extracted_data = []
```

```python
city_name = city.split(",")[0].strip().lower()

for place in places:
    tag = {"amenity": place}

    tagged_data = osmnx.features_from_place(city, tags = tag)
    file_name = f"{place}_{city_name}.xlsx"
    path = join(DATA_PATH, file_name)

    # merge_cells=False: otherwise Excel visually merges repeated (element, id) index
    # values and reading it back leaves those rows blank instead of repeating the value.
    tagged_data.assign(geometry=tagged_data["geometry"].apply(shapely.to_wkt)).to_excel(path, merge_cells=False)
    extracted_data.append(tagged_data)
```

```python
all_data = pd.concat(extracted_data, ignore_index=True)
if "cuisine" in all_data.columns:
    print(all_data["cuisine"].value_counts(dropna=False))
else:
    print("No cuisine tags found in this pull.")
```
